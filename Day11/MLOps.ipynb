```python
mapping = {
    'zero': 0,
    'one': 1,
    'two': 2,
    'three': 3,
    'four': 4,
    'five': 5,
    'six': 6,
    'seven': 7,
    'eight': 8,
    'nine': 9,
    'ten': 10
}
```

```python
df['experience'] = df['experience'].astype(str).str.lower()
df['experience'] = df['experience'].map(mapping)
```

```python
df['test_score'] = df['test_score'].fillna(df['test_score'].mean())
```

```python
df['experience'] = df['experience'].fillna(0)
```

```python
df
```

```python
x=df.drop('salary',axis=1)
y=df[['salary']]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
model=lr.fit(x,y)
```

```python
import pickle
```

```python
pickle.dump(model,open('maas.pkl','wb'))
```

```python
kmodel=pickle.load(open('maas.pkl','rb'))
```

```python
kmodel.predict([[5,8,9]])[0][0]
```
